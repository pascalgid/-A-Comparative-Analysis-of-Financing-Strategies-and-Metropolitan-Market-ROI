```python
from pathlib import Path
import numpy as np

# Question 1: Read the CSV file into a matrix named Iris.
data_file = Path("ML_HW_Data_FisherIris.csv")
Iris = np.loadtxt(data_file, delimiter=",")

# Question 2: Display the original number of rows and columns.
print(f"2. Original Iris matrix: {Iris.shape[0]} rows and {Iris.shape[1]} columns")

# Question 3: Display the record numbers whose fifth-column value is negative.
# Add 1 because Python array indices start at 0, while record numbers start at 1.
negative_row_numbers = np.where(Iris[:, 4] < 0)[0] + 1
print(
    "3. Row numbers with a fifth-column value less than 0: "
    f"{negative_row_numbers.tolist()}"
)

# Question 4: Remove rows whose fifth-column value is negative.
Iris = Iris[Iris[:, 4] >= 0]

# Question 5: Display the new number of rows and columns.
print(f"5. Cleaned Iris matrix: {Iris.shape[0]} rows and {Iris.shape[1]} columns")

# Question 6: Copy the first four columns into X.
X = Iris[:, :4]

# Question 7: Copy the fifth column into Y.
Y = Iris[:, 4]

# Question 8: Display the maximum and minimum of each column in X.
column_maximums = X.max(axis=0)
column_minimums = X.min(axis=0)
print("8. Maximum and minimum values for each column in X:")
for column_number, (maximum, minimum) in enumerate(
    zip(column_maximums, column_minimums), start=1
):
    print(
        f"   Column {column_number}: maximum = {maximum:g}, "
        f"minimum = {minimum:g}"
    )

# Question 9: Count values greater than 36 in the third column of X.
third_column_count = np.count_nonzero(X[:, 2] > 36)
print(f"9. Number of elements in column 3 of X greater than 36: {third_column_count}")
```
